```python
features_names = ["pregnant", "glucose", "bp", "skin", "insulin", "bmi", "pedigree", "age", "flag"]
```

```python
df_mod_diabets = pd.read_csv("pima-indians-diabetes.data", header=None, names=features_names)
```

```python
#View dataset
df_mod_diabets.head()
```

```python
df_mod_diabets.shape
```

```python
features_select_df = ["pregnant", "insulin", "bmi", "age"]
```

```python
#Creating featers object
x_features = df_mod_diabets[features_select_df]
```

```python
y_target = df_mod_diabets["flag"]
```

```python
x_features
```

```python
y_target
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_features, y_target, random_state=27)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_reg = LogisticRegression()
```

```python
log_reg.fit(x_train, y_train)
```

```python
#prediction user dataset
y_pred = log_reg.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(y_test, y_pred))
```
