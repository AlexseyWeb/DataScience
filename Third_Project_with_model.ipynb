### Create to model and testing 

```python
x_features = df_mod[['TV Ad Budget ($)', 'Radio Ad Budget ($)', 'Newspaper Ad Budget ($)']]
```

```python
x_features.head()
```

```python
#Create a target object
y_target = df_mod[['Sales ($)']]
```

```python
y_target.head()
```

```python
x_features.shape
```

```python
y_target.shape
```

#### Split dataset on traning and test data 

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_features, y_target, random_state=27)
```

```python
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

# Create a linear regression  model

```python
from sklearn.linear_model import LinearRegression
```

```python
linreg = LinearRegression()
```

```python
linreg.fit(x_train, y_train)
```

```python
print(linreg.intercept_)
```

```python
print(linreg.coef_)
```

```python
y_pred = linreg.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

# Score MSE

```python
from sklearn import metrics
import numpy as np
```

```python
#Calculate MSE
print(metrics.mean_squared_error(y_test, y_pred))
```

```python
#Calculate RMSE
print(metrics.root_mean_squared_error(y_test, y_pred))
```

```python
# NUMPY version RMSE
print(np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
```
